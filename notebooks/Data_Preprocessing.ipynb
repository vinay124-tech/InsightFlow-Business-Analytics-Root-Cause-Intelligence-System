```python
import pandas as pd
```

```python
df = pd.read_csv("../Data/interim/retail_merged.csv")
df.head()
```

```python
df.shape
```

```python
df = df.drop_duplicates()
df.shape
```

```python
cancelled = df["Invoice"].astype(str).str.startswith("C")
cancelled.shape
```

```python
df = df[~cancelled]
df.shape
```

```python
print(f"Removing {(df['Quantity'] <= 0).sum():,} invalid quantities...")
```

```python
df = df[df["Quantity"] > 0]
df.shape
```

```python
print(f"Removing {(df['Price'] <= 0).sum():,} invalid prices...")
```

```python
df = df[df["Price"] > 0]
df.shape
```

```python
print(f"Removing {df['Description'].isna().sum():,} missing descriptions...")
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df.head()
```

```python
df["Month"] = df["InvoiceDate"].dt.month
df["MonthName"] = df["InvoiceDate"].dt.month_name()
df["Quarter"] = df["InvoiceDate"].dt.quarter
df["Day"] = df["InvoiceDate"].dt.day
df["DayName"] = df["InvoiceDate"].dt.day_name()
df["Hour"] = df["InvoiceDate"].dt.hour
df["Weekday"] = df["InvoiceDate"].dt.weekday
```

```python
df.head()
```

```python
df.shape
```

```python
df["Revenue"] = df["Quantity"] * df["Price"]
```

```python
print("\n" + "="*80)
print("FINAL DATASET")
print("="*80)

print(f"Final Shape : {df.shape}")

print("\nMissing Values")
display(df.isnull().sum())

print("\nData Types")
display(df.dtypes)

print("\nRevenue")
print(f"Total Revenue : {df['Revenue'].sum():,.2f}")

display(df.head())
```

### Saving the processed Data

```python
df.to_csv("../Data/processed/final_retail.csv", index=False)
```
