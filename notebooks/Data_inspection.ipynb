```python
# ==========================================================
# COMPLETE DATASET INSPECTION
# ==========================================================

import pandas as pd

print("="*80)
print("DATASET OVERVIEW")
print("="*80)

print(f"Rows               : {retail_df.shape[0]:,}")
print(f"Columns            : {retail_df.shape[1]}")
print(f"Date Range         : {retail_df['InvoiceDate'].min()}  -->  {retail_df['InvoiceDate'].max()}")

print("\nColumn Names:")
print(retail_df.columns.tolist())

# ==========================================================
print("\n" + "="*80)
print("DATA TYPES & MEMORY")
print("="*80)

retail_df.info()

# ==========================================================
print("\n" + "="*80)
print("NUMERICAL SUMMARY")
print("="*80)

display(retail_df.describe().T)

print("\n" + "="*80)
print("CATEGORICAL SUMMARY")
print("="*80)

display(retail_df.describe(include="object").T)

# ==========================================================
print("\n" + "="*80)
print("MISSING VALUES")
print("="*80)

missing = pd.DataFrame({
    "Missing Values": retail_df.isnull().sum(),
    "Percentage (%)": (retail_df.isnull().mean()*100).round(2)
})

display(missing.sort_values("Missing Values", ascending=False))

# ==========================================================
print("\n" + "="*80)
print("DUPLICATE ROWS")
print("="*80)

duplicates = retail_df.duplicated().sum()
print(f"Duplicate Rows : {duplicates:,}")

# ==========================================================
print("\n" + "="*80)
print("UNIQUE VALUES")
print("="*80)

display(pd.DataFrame({
    "Unique Values": retail_df.nunique()
}).sort_values("Unique Values"))

# ==========================================================
print("\n" + "="*80)
print("BUSINESS QUALITY CHECKS")
print("="*80)

cancelled = retail_df["Invoice"].astype(str).str.startswith("C").sum()

negative_qty = (retail_df["Quantity"] < 0).sum()
zero_qty = (retail_df["Quantity"] == 0).sum()

negative_price = (retail_df["Price"] < 0).sum()
zero_price = (retail_df["Price"] == 0).sum()

missing_customers = retail_df["Customer ID"].isnull().sum()

print(f"Cancelled Invoices     : {cancelled:,}")
print(f"Negative Quantity      : {negative_qty:,}")
print(f"Zero Quantity          : {zero_qty:,}")
print(f"Negative Price         : {negative_price:,}")
print(f"Zero Price             : {zero_price:,}")
print(f"Missing Customer IDs   : {missing_customers:,}")

# ==========================================================
print("\n" + "="*80)
print("CUSTOMER INSIGHTS")
print("="*80)

print(f"Unique Customers : {retail_df['Customer ID'].nunique():,}")

# ==========================================================
print("\n" + "="*80)
print("PRODUCT INSIGHTS")
print("="*80)

print(f"Unique Stock Codes         : {retail_df['StockCode'].nunique():,}")
print(f"Unique Product Descriptions: {retail_df['Description'].nunique():,}")

# ==========================================================
print("\n" + "="*80)
print("COUNTRY INSIGHTS")
print("="*80)

print(f"Number of Countries : {retail_df['Country'].nunique()}")

display(retail_df["Country"].value_counts().head(20))

# ==========================================================
print("\n" + "="*80)
print("TOP PRODUCTS")
print("="*80)

display(retail_df["Description"].value_counts().head(20))

# ==========================================================
print("\n" + "="*80)
print("REVENUE")
print("="*80)

retail_df["Revenue"] = retail_df["Quantity"] * retail_df["Price"]

print(f"Total Revenue (Raw Dataset): {retail_df['Revenue'].sum():,.2f}")

display(retail_df[["Quantity","Price","Revenue"]].head())

# ==========================================================
print("\n" + "="*80)
print("RANDOM SAMPLE")
print("="*80)

display(retail_df.sample(10, random_state=42))
```

```python
output_path = "../data/interim/retail_merged.parquet"

try:
    retail_df.to_parquet(output_path, index=False)
    print("Merged dataset saved successfully!")
except ImportError:
    fallback_path = output_path.replace(".parquet", ".csv")
    retail_df.to_csv(fallback_path, index=False)
    print(f"Parquet engine not available. Saved as CSV instead: {fallback_path}")

print("Merged dataset saved successfully!")
```
